```python
import json
import re
import tiktoken
from tqdm import tqdm
```

```python
with open("StdEnrich.json", "r") as rfile:
    data = json.load(rfile)
```

```python
pattern_homo = re.compile(r"\b(?:human|humans|homo\s+sapiens)\b", flags=re.IGNORECASE)
pattern_mus = re.compile(r"\b(?:mouse|mice|murine|mus\s+musculus)\b", flags=re.IGNORECASE)
for key, value in tqdm(data.items()):
    species = []
    claim = value["Applicable species"]
    # matches_homo = pattern_homo.findall(claim)
    # matches_mus = pattern_mus.findall(claim)
    
    # if matches_homo and matches_mus:
    #     species.append("Human")
    #     species.append("Mouse")
    # elif matches_homo and not matches_mus:
    #     species.append("Human")
    # elif not matches_homo and matches_mus:
    #     species.append("Mouse")
    # else:
    #     species.append("Unknown")
    for term in claim:
        if term == "Human":
            term = term.replace("Human", "Homo sapiens")
            species.append(term)
        if term == "Mouse":
            term = term.replace("Mouse", "Mus musculus")
            species.append(term)
            
    value["Applicable species"] = species
    
with open("StdEnrich.json", "w") as wfile:
    json.dump(data, wfile, indent=4)
```

```python
new_data = {}
for key, value in data.items():
    narrow_value = value["Overall functional summary"] + " " + value["What the gene sets represent"]
    new_data[key] = narrow_value
print(len(new_data))
with open("../document/StdLibrary.json", "w") as wfile:
    json.dump(new_data, wfile, indent=4, ensure_ascii=False)
```
